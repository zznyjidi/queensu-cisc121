```python
def max_sum_o3(array):
    n = len(array)
    best = float('-inf')
    
    for a in range(n):
        for b in range(a, n):
            array_sum = 0
            for k in range(a, b+1):
                array_sum += array[k]
            best = max(best, array_sum)
    return best
```

```python
def kadane(array):
    best = 0
    array_sum = 0
    for i in array:
        array_sum = max(i, array_sum + i)
        best = max(best, array_sum)
    return best
```

```python
import time
```

```python
def benchmark_func(func, *args, **kwargs) -> float:
    start = time.perf_counter()
    func(*args, **kwargs)
    end = time.perf_counter()
    return end - start
```

```python
import random

import datascience

random.seed('2c5c90d1d269e044836285b80e17470daa38c2c0321812e0d758dbed98773442')
```

```python
def random_array(length, start=-20, stop=20):
    return [random.randint(start, stop) for _ in range(length)]
```

```python
def benchmark(func, count, length):
	runtime = datascience.Table([i.__name__ for i in func])
	for _ in range(count):
		a = random_array(length)
		runtime.append([benchmark_func(i, a) for i in func])
	return runtime
```

```python
benchmark([max_sum_o3, max_sum_o2, kadane], 20, 1000).stats()
```

```python
benchmark([max_sum_o2, kadane], 20, 10000).stats()
```
